```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/churn_clean_raw.csv")
df.shape
```

```python
df["Churn Value"].value_counts(normalize=True) * 100
```

```python
figure = plt.figure(figsize=(4, 4))
sns.countplot(data=df, x="Churn Value")
plt.title("Churn Distribution")
plt.show()
```

```python
df.groupby("Contract")["Churn Value"].mean().sort_values(ascending=False)
```

```python
figure = plt.figure(figsize=(5, 4))
sns.barplot(data=df, x="Contract", y="Churn Value")
plt.title("Churn Rate by Contract Type")
plt.ylabel("Churn Rate")
plt.show()
```

```python
figure = plt.figure(figsize=(5, 4))
sns.barplot(data=df, x="Internet Service", y="Churn Value")
plt.title("Churn Rate by Internet Service")
plt.ylabel("Churn Rate")
plt.show()
```

```python
figure = plt.figure(figsize=(7, 6))
sns.histplot(data=df, x="Tenure Months", hue="Churn Value", multiple="stack", bins=30)
plt.title("Tenure Distribution by Churn")
plt.show()
```

```python
figure = plt.figure(figsize=(4, 6))
sns.boxplot(data=df, x="Churn Value", y="Monthly Charges")
plt.title("Monthly Charges by Churn")
plt.show()
```

```python
figure = plt.figure(figsize=(8, 8))
numeric_df = df.select_dtypes(include="number")
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="PuBu", cbar=True, square=True)
plt.title("Correlation Matrix")
plt.show()
```

##### Takeaways
Month-to-month contracts churn far more than one/two year contracts
Fiber optic customers churn more than DSL or no-internet customers
Churn is concentrated in the first ~12 months of tenure
Higher monthly charges are associated with higher churn
